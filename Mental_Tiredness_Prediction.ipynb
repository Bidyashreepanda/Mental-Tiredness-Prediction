**Extract the numerical , nominal and ordinal column**

```python
numerical_columns = [
    "number_of_decisions_made",
    "context_switch_count",
    "notifications_received",
    "screen_time_min",
    "deep_work_min",
    "task_complexity_avg",
    "caffeine_mg",
    "break_frequency",
    "sleep_hours",
    "deep_sleep_pct",
    "hydration_l",
    "noise_level_db",
    "temperature_c",
    "workload_score"
]

ordinal_columns = [
    "mood",
    "work_environment"
]

nominal_columns = [
    "work_type"
]
```

**Create the numerical,nominal and ordinal pipeline**

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder())
])
```

```python
nominal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

**Column Transformer**

```python
preprocessor = ColumnTransformer([
    ("num", num_pipeline, numerical_columns),
    ("ordinal", ordinal_pipeline, ordinal_columns),
    ("nominal", nominal_pipeline, nominal_columns)
])
```

**Fit Transform on x_train and x_test**

```python
X_test = preprocessor.transform(x_test)
```

**Sprint 2 : Model Evaluation**

```python
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

We imported the required Scikit-learn libraries to train regression models and evaluate their performance using standard regression metrics.

LinearRegression: Builds a linear regression model.



DecisionTreeRegressor: Builds a decision tree model.



RandomForestRegressor: Builds multiple decision trees.



GradientBoostingRegressor: Builds models sequentially to improve predictions.



sklearn.metrics: Measures model performance.

**Define the models**

```python
models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    )
}
```

Why do we train multiple models?

Different algorithms learn different patterns from data. Training multiple models helps us identify which algorithm predicts mental tiredness most accurately.

Insight to write:

Four regression algorithms were selected to compare their predictive performance and identify the most suitable model for the dataset.

**Create a preprocessing + model pipeline**

```python
results = []

trained_models = {}

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipeline.fit(x_train, y_train)

    trained_models[name] = pipeline

    print(name, "training completed")
```

A machine learning pipeline was created to combine preprocessing and model training. This ensures consistent transformations and prevents data leakage by fitting preprocessing steps only on the training data.

**Make Predictions**

```python
y_train_pred = pipeline.predict(x_train)

y_test_pred = pipeline.predict(x_test)

print("Training Predictions:", y_train_pred[:5])
print("Testing Predictions:", y_test_pred[:5])
print()
```

All four trained models generated predictions on both training and testing datasets to evaluate their ability to learn patterns and predict unseen observations.

**Evaluate Model Performance**

```python
y_train_pred = pipeline.predict(x_train)
y_test_pred = pipeline.predict(x_test)
```

```python
mae = mean_absolute_error(y_test, y_test_pred)
print("MAE:", mae)
```

```python
mse = mean_squared_error(y_test, y_test_pred)
print("MSE:", mse)
```

```python
rmse = np.sqrt(mse)
print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_test_pred)
print("R2 Score:", r2)
```

The four regression models were evaluated using MAE, MSE, RMSE, and R². Lower error values and a higher R² score indicate better predictive performance.

**Create Model Comparison table**

```python
results = []

for name, pipeline in trained_models.items():

    y_train_pred = pipeline.predict(x_train)
    y_test_pred = pipeline.predict(x_test)

    results.append({
        "Model": name,

        "Train MAE": mean_absolute_error(
            y_train, y_train_pred
        ),

        "Test MAE": mean_absolute_error(
            y_test, y_test_pred
        ),

        "Test MSE": mean_squared_error(
            y_test, y_test_pred
        ),

        "Test RMSE": np.sqrt(
            mean_squared_error(y_test, y_test_pred)
        ),

        "Train R2": r2_score(y_train, y_train_pred),

        "Test R2": r2_score(y_test, y_test_pred)
    })

comparison_df = pd.DataFrame(results)

comparison_df
```

A model comparison table was created to compare the performance of all four regression algorithms. The model with lower prediction errors and better generalization performance will be considered the best baseline candidate.

**Check overfitting and underfitting**

```python
for index, row in comparison_df.iterrows():

    print("Model:", row["Model"])

    print("Training R2:", row["Train R2"])

    print("Testing R2:", row["Test R2"])

    difference = row["Train R2"] - row["Test R2"]

    print("Difference:", difference)

    if difference > 0.15:
        print("Possible Overfitting")

    elif row["Train R2"] < 0.50:
        print("Possible Underfitting")

    else:
        print("Check for reasonable generalization")

    print("-" * 40)
```

Training and testing performance were compared to identify overfitting and underfitting. A smaller performance gap generally indicates better generalization, provided the model's overall performance is also good.

**SPRINT 3: OPTIMIZATION & FINAL MODEL**

**Feature Engineering**

```python
# 1. Screen time compared with sleep time
df["screen_sleep_ratio"] = (
    df["screen_time_min"] / (df["sleep_hours"] * 60 + 1)
)
```

```python
# 2. Deep work compared with total screen time
df["deep_work_ratio"] = (
    df["deep_work_min"] / (df["screen_time_min"] + 1)
)
```

```python
# 3. Notifications per hour of deep work
df["notification_load"] = (
    df["notifications_received"] /
    (df["deep_work_min"] / 60 + 1)
)
```

```python
# Check the new features
df[[
    "screen_sleep_ratio",
    "deep_work_ratio",
    "notification_load"
]].head()
```

1. Created new features using existing lifestyle and work-related variables.
2. Screen-to-sleep ratio helps understand the relationship between screen exposure and sleep duration.
3. Deep work ratio provides information about focused work compared with screen time.
4. Notification load helps measure interruptions relative to deep work duration.
5. These engineered features may improve the predictive performance of the model.

**Recursive Feature Elimination**

```python
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestRegressor

# Use numerical features only for this RFE demonstration
rfe = RFE(
    estimator=RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    n_features_to_select=8
)

rfe.fit(x_train[numerical_columns], y_train)

selected_features = x_train[numerical_columns].columns[
    rfe.support_
]

print("Selected Features:")
print(selected_features.tolist())
```

4. Recursive Feature Elimination was applied to identify a smaller subset of numerical features.
5. Feature selection aims to improve model efficiency and reduce unnecessary complexity.

**Hyperparameter Tuning**

```python
#Grid Search CV
param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [None, 10, 20],
    "model__min_samples_split": [2, 5]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
#create the pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
#create the parameter grid
param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [None, 10, 20],
    "model__min_samples_split": [2, 5],
    "model__min_samples_leaf": [1, 2]
}
```

```python
#apply Grid Search cv
grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    verbose=1
)

grid_search.fit(x_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("Best CV RMSE:")
print(-grid_search.best_score_)
```

1. Hyperparameter tuning was performed to improve the model's predictive performance.



2. GridSearchCV evaluated different parameter combinations using 5-fold cross-validation.



3. The best parameter combination was identified based on cross-validation RMSE.



4. Cross-validation helped assess model performance across different training and validation folds.



5. The best hyperparameters will be used to build the final model.

**Build the Final Model**

```python
final_model = grid_search.best_estimator_

print("Final Model:")
print(final_model)

print("\nBest Parameters:")
print(grid_search.best_params_)
```

```python
print("Final model is ready!")
```

1. The best-performing hyperparameter combination was selected using GridSearchCV.
2. The final model includes both preprocessing and regression steps.
